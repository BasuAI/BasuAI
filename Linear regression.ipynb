# **1. Prepararion of DATASET**

## **Load Dataset**

```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/BushraAijaz/dataset/main/DummyData1009.csv')
df
```

## **Separation of data as x and y**

```python
y = df['output_y']
x = df.drop('output_y', axis = 1)
```

## **Spliting of DAtaset into Training and Testing**

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size =0.2, random_state =100)
```

# **2. MODEL BUILDING -- Linear Regression**

## **Model Training**

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)
```

## **Applying a model to make a prediction**

```python
lr_train_y_pred = lr.predict(x_train)
lr_test_y_pred =lr.predict(x_test)
lr_test_y_pred
```

# **3. Evaluate Model Performance**

```python
from sklearn.metrics import mean_squared_error, r2_score
lr_train_MSE = mean_squared_error(y_train, lr_train_y_pred)
lr_train_R2 = r2_score(y_train,lr_train_y_pred)
lr_test_MSE = mean_squared_error(y_test, lr_test_y_pred)
lr_test_R2 = r2_score(y_test, lr_test_y_pred)
```

```python
lr_results = pd.DataFrame(['Linear Regression',lr_train_MSE, lr_train_R2, lr_test_MSE,lr_test_R2 ]).transpose()
lr_results.columns = ['Methods', 'Taining MSE', 'Training r2', 'Test MSE', 'Test R2']
lr_results
```

# **4. Data Visualization of Prediction Results**

```python
import matplotlib.pyplot as plt
import numpy as np
plt.figure(figsize=(5,5))
plt.scatter(x=y_train, y=lr_train_y_pred, c='g',alpha=0.3 )
z=np.polyfit(y_train, lr_train_y_pred, 1)
p=np.poly1d(z) #the 1D poly class
plt.plot(y_train, p(y_train), 'r')
plt.ylabel('Predict output_y')
plt.xlabel('Experimental output_y')
```
